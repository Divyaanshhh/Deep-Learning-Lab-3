## Feedforward Neural Network for Loan Approval Prediction

```python
import numpy as np
import tensorflow as tf
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Sample Dataset
# Features:
# Credit Score (0-900)
# Annual Income (in Lakhs)
# Employment Status (0=Unemployed/Self-employed, 1=Salaried)

X = np.array([
    [350, 2.5, 0],
    [720, 8.0, 1],
    [800, 12.0, 1],
    [400, 3.0, 0],
    [680, 7.5, 1],
    [450, 4.0, 0],
    [750, 10.0, 1],
    [380, 2.8, 0]
], dtype=float)

y = np.array([0, 1, 1, 0, 1, 0, 1, 0])
```

```python
# Normalize Numerical Features
X[:,0] = X[:,0] / 900    # Credit Score (max 900)
X[:,1] = X[:,1] / 15     # Annual Income (max assumed 15 Lakhs)
```

```python
# Build Feedforward Neural Network
model = Sequential([
    Dense(8, activation='relu', input_shape=(3,)),
    Dense(4, activation='relu'),
    Dense(1, activation='sigmoid')
])
```

```python
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)
```

```python
# Train Model
model.fit(X, y, epochs=100, verbose=0)
```

```python
# Evaluate Model
loss, accuracy = model.evaluate(X, y, verbose=0)
print('Accuracy:', accuracy)
```

```python
# Predictions
predictions = model.predict(X)

print('\nLoan Approval Predictions:')
for i, pred in enumerate(predictions):
    result = 1 if pred[0] >= 0.5 else 0
    status = 'Approved' if result == 1 else 'Rejected'
    print(f'Customer {i+1}: Credit={int(X[i][0]*900)}, Income={X[i][1]*15:.1f}L, Salaried={int(X[i][2])} -> {status} (Confidence: {pred[0]:.3f})')
```
